# Homework 1 Gaussian Distributions 

```python
# importing needed packages
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
from scipy import stats

#this sets the size of the plot to something useful
plt.rcParams["figure.figsize"] = (15,10)
```

```python
#creating array with points picked from normal distribution with mean at 2.0 and standard deviation of .05
d = stats.norm.rvs(loc = 2.0, scale = .05, size = 1000000)
```

The points in the array d are chosen from a Gaussian distribution
$$ f(x) = \frac{1}{\sqrt{2\pi\sigma^2}}e^\frac{-(x-\mu)^2}{2\sigma^2} $$
Where the mean equals $$ \sigma = .05 $$ and the standard deviation equals$$ \mu = 2 $$

```python
#the points in d are plotted using a histogram
ax = plt.hist(d,100, density=True)
plt.tick_params(labelsize = 24)
plt.xlim([1.75,2.25])
plt.show()
```

```python
#the points in d are plotted again this time using a log scale for the y-axis
ax = plt.hist(d,100, density=True)
plt.yscale('log')
plt.tick_params(labelsize = 24)
plt.xlim([1.75,2.25])
plt.show()
```

Plotting the data using a log scale allows us to see the tail of the distribution. It is noted that a gaussian will follow a parabolic curve when plotted with a semi-log plot. The above distribution appears to follow such a shape, though the tails of the histogram do not. Such features likely occur due to insufficient sampling of the tails of the distribution in the data used rather than due to the nature of the distrubtion the data is sampled from. 

```python
#the data is plotted again, this time with a gaussian curve of the same distribution plotted alongside the histogram from before
fig, ax = plt.subplots(1, 1)
ax.hist(d,100, density=True)
plt.yscale('log')
plt.tick_params(labelsize = 24)
plt.xlim([1.75,2.25])
x = np.linspace(1.75,2.25,1000)
ax.plot(x,stats.norm.pdf(x,loc = 2., scale = 0.05),linewidth = 8,alpha = 0.7)
plt.show()
```
